```python
from langgraph.store.memory import InMemoryStore
```

```python
# Creating a store
store = InMemoryStore()
```

```python
# Creating a namespace (Always in tuple of strings)
namespace = ("user","u1")
```

# Creating Memories

```python
# Adding memories
store.put(namespace,"1",{"data" : "user likes pizza"})
store.put(namespace,"2",{"data" : "user prefers dark mode"})
```

```python
namespace2 = ("user","u2")

# Adding memories
store.put(namespace2,"1",{"data" : "User likes bmw cars"})
store.put(namespace2,"2",{"data" : "User likes rajma chawal"})
```

# Retrieving Memories

```python
# store.get(namespace,key)
store.get(namespace,"1")
```

# Retrieving All Memories

```python
items = store.search(namespace)

for item in items:
    print(item.value)
```
